## Section 4 — Validation vs KDDTest+ generalization

Absolute change = KDDTest+ Macro F1 − validation Macro F1.  
Relative change (%) = (KDDTest+ − validation) / validation × 100.

Class-weighted Logistic Regression has **no completed KDDTest+ row in this comparison set** (Day 4/5 final tables used no-SMOTE, SMOTE, and Random Forest). That row is omitted rather than invented.

```python
gen_rows = []
for name in [
    "Logistic Regression — no SMOTE",
    "Logistic Regression — SMOTE",
    "Random Forest — 500 trees",
]:
    val_f1 = float(
        day7_validation_comparison.loc[
            day7_validation_comparison["Evaluation"] == name, "Macro F1"
        ].iloc[0]
    )
    test_f1 = float(
        day7_kddtest_comparison.loc[
            day7_kddtest_comparison["Evaluation"] == name, "Macro F1"
        ].iloc[0]
    )
    abs_change = test_f1 - val_f1
    rel_change = abs_change / val_f1 * 100
    gen_rows.append(
        {
            "Evaluation": name,
            "Validation Macro F1": val_f1,
            "KDDTest+ Macro F1": test_f1,
            "Absolute Macro F1 Change": abs_change,
            "Relative Macro F1 Change (%)": rel_change,
        }
    )

day7_generalization_macro_f1 = pd.DataFrame(gen_rows)
print("Generalization comparison (not used for model selection)")
display(day7_generalization_macro_f1.round(4))
```

## Section 7 — Why Random Forest remains locked

Random Forest had the **strongest validation Macro F1 (0.951031)** among the four reported validation candidates. It was therefore selected under the **predefined validation-only** procedure, **before** KDDTest+ was scored.

KDDTest+ was **held out** from selection. Logistic Regression with SMOTE has a **higher observed KDDTest+ Macro F1 (0.5863 vs 0.5061)**. That fact **cannot retroactively change** the lock. Using the held-out test set to pick a different final model would be **test-set selection**.

Random Forest 500 is **not** claimed to be the best model **on KDDTest+**. It is the model **selected by the validation-based experimental procedure**. It remains the locked final model.

## Section 8 — Overall experimental conclusions

1. Random Forest substantially outperformed Logistic Regression on **KDDTrain+ validation Macro F1** (0.951031 vs 0.856000 no-SMOTE, 0.702800 SMOTE, 0.698500 class-weighted).
2. All three models with both validation and KDDTest+ scores show a **validation-to-KDDTest+ generalization gap** (Macro F1 falls in every case).
3. **Logistic Regression with SMOTE** had the **highest observed KDDTest+ Macro F1** among the evaluated models (0.5863).
4. **Random Forest 500 remains the locked final model** because selection occurred **before** KDDTest+ evaluation.
5. **Rare-class** performance, especially **R2L and U2R**, deteriorated substantially on KDDTest+ for the locked forest (and declined for the Logistic Regression models as well).
6. Day 6 analysis provided **evidence consistent with distribution/domain shift**; it did not prove that any one feature caused the errors.
7. These KDDTest+ numbers are an **evaluation of generalization**. They are **not** a reason to tune or re-select a model on the test set.
