## Section 8 — Final project conclusion

1. Random Forest 500 achieved **excellent validation** performance (Macro F1 = 0.951031).
2. It was **selected before** KDDTest+ evaluation.
3. Its **KDDTest+ performance was substantially lower** (Macro F1 = 0.5061).
4. **R2L and U2R generalization** were the major weaknesses.
5. KDDTest+ has a **substantially different class composition** (R2L 0.7898% → 12.7972%; U2R 0.0397% → 0.2972%).
6. Numerical and class-conditional distribution shifts provide **evidence consistent with dataset/domain shift**.
7. Those shifts **help explain** the validation-to-test gap but **do not prove causality**.
8. Random Forest **remains the final model** because selection followed the **predefined validation procedure**, not KDDTest+ scores.
